# Preprocessing

Carregamento e processamento inicial dos dados.

```python
import pandas as pd
import numpy as np
```

```python
# Carregar os dados brutos
df = pd.read_excel('raw/GulfCoast_NMR.xlsx')
df.head()
```

```python
# Selecionar entre as linhas 960-1536
df_filtered = df.loc[960:1534].copy()
```

- $$S_{wirr} = \frac{MBVI}{PHIX}$$

```python
# Criar duas colunas de saturação de água irredutível
df_filtered['Swirr_PHIX'] = df_filtered['MBVI'] / df_filtered['PHIX']
df_filtered.head()
```
